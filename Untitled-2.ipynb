```python
#%pip install pandas
#%pip install matplotlib

import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
x = [2023, 2024, 2025, 2026]
y1 = [23, 42, 40, 34]
y2 = [39, 41, 23, 18]
y3 = [31, 10, 9, 39]
```

```python
line_style = dict(marker=".",
         markersize=2,
         markerfacecolor="#47b1b5",
         markeredgecolor="#17c1d5",
         linestyle="dotted",
         linewidth=3)
plt.title("Class size", fontsize=20, family="Arial", fontweight="bold", color="darkred")



plt.plot(x,y1, color="#fbff00", **line_style) #ms, mfc or mec could be used as well
plt.plot(x, y2, color="#ed2341", **line_style)
plt.plot(x, y3, color="#003cff", **line_style)
plt.tick_params(axis="both", colors="#00eeff")
plt.xticks(x)
plt.show()
```

```python
x = [1,2,3,4,5]
y=[5,10,15,20,25]
plt.grid(axis="x",
         linewidth=1)
plt.plot(x,y)
plt.show()
```

```python
#Barchart
categories = ["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"]
values = [4, 3, 2, 5, 3, 1]

plt.bar(categories, values)
```

```python
plt.pie(values, labels=categories,
        autopct="%1.1f%%",
        explode=[0,0,0.1,0,0,0] )
```

```python
## Subplot
# Figure = The entrie canvas
# Ax = A single plot (subplot)
figure, axes = plt.subplots(2,2)


x = np.array([1,2,3,4,5])
axes[0,0].plot(x,x*2,color="red")
axes[0,0].set_title("x*2")

axes[0,1].plot(x, x**2, color="blue")
axes[0,1].set_title("x**2")

axes[1,0].plot(x**3, color="green")
axes[1,0].set_title("x**3")

axes[1,1].plot(x,x**4, color="purple")
axes[1,1].set_title("x**4")

plt.tight_layout()
plt.show()
```
